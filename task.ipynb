```python
import pandas as pd
from datetime import timedelta

df = pd.read_csv('source.csv')
```

column 'Datetime' UTC to UTC+6

```python
df['Datetime'] = pd.to_datetime(df['Datetime']) + timedelta(hours=6)
```

```python
df.head(1)
```

Datetime is "Datetime"

```python
if not pd.api.types.is_datetime64_any_dtype(df['Datetime']):
    df['Datetime'] = pd.to_datetime(df['Datetime'])
```

```python
def calculate_total(row, df):
    purity_factor = 0.75 if row['Purity'] == 'Impure' else 1.0

    if row['Name'] == 'ProductA':
        return row['Amount'] * row['Price'] * purity_factor

    elif row['Name'] == 'ProductB':
        # Search price of ProductA with the same Datetime
        matching_a = df[(df['Name'] == 'ProductA') & (df['Datetime'] == row['Datetime'])]
        if not matching_a.empty:
            price_a = matching_a.iloc[0]['Price']
            diff_price = row['Price'] - price_a
            return row['Amount'] * diff_price * purity_factor
        else:
            return 0
    else:
        return 0
```

```python
df['total'] = df.apply(lambda row: calculate_total(row, df), axis=1)
```

```python
df.to_csv('result.csv', index=False)
```
